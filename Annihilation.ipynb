```python
# Cell 3: start a game and show the board
import quantum as Q
from game import GameState, play, HELP

Q.set_seed(7)
game = GameState()
game.show()
print()
print(HELP)
```

```python
# Cell 4: the quantum circuits behind the game
circuits = {
    "Fork (50/50)": Q.fork_circuit(),
    "Fork after Interference (fixed to B)": Q.fork_circuit(1),
    "Power-up draw (QRNG)": Q.qrng_circuit(),
    "Bell pair": Q.bell_circuit(),
    "Teleport check": Q.teleport_circuit(1.0, 2.0),
}
for name, qc in circuits.items():
    print(name, Q.counts(qc, shots=1000))
    display(qc.draw("mpl"))
```

```python
# Cell 5: a whole game played with random legal commands
import random

rng = random.Random(3)
demo = GameState()
while not demo.over:
    demo.command(rng.choice(demo.legal_commands()))
print("\n".join(demo.log[-12:]))
print()
demo.show()
```
